```python
folder_name = r"Garbage_classification"
file_path = os.path.join(output_dir, folder_name)

base_dataset = ImageFolder(root=file_path, transform=None)

class_dict = dict(enumerate(base_dataset.classes))

print(f"총 데이터수: {len(base_dataset)}")
print(f"클래스 수: {len(class_dict)}")
print(f"라벨 매핑: {class_dict}")
```

```python
mean = [0.485, 0.456, 0.406]
std = [0.229, 0.224, 0.225]

train_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ToTensor(),
    transforms.Normalize(mean=mean, std=std)
])

val_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=mean, std=std)
])
```

```python
# Dataset객체를 넘기면 sklearn이 인덱싱을 하기위해 base_dataset 0~2527까지 전부 호출하게 되는데
# 시간도 오래걸리고 튜플에 담긴 리스트로 나오게 된다.

# train_datas, test_datas = train_test_split(
#     base_dataset,
#     test_size=0.2,
#     shuffle=True,
#     random_state=42
# )

train_idx, val_idx = train_test_split(
    range(len(base_dataset)),
    test_size=0.2, 
    stratify=base_dataset.targets, 
    random_state=42
)

train_subset = Subset(base_dataset, train_idx) 
val_subset = Subset(base_dataset, val_idx)

len(train_subset), len(val_subset)
```

```python
class DatasetWrapper(Dataset):
    def __init__(self, dataset, transform=None):
        self.dataset = dataset
        self.transform = transform

    def __len__(self):
        return len(self.dataset)

    def __getitem__(self, idx):
        x, y = self.dataset[idx]
        if self.transform:
            x = self.transform(x)
        return x, y
```

```python
train_dataset = DatasetWrapper(train_subset, transform=train_transform)
val_dataset = DatasetWrapper(val_subset, transform=val_transform)

train_dataloader = DataLoader(train_dataset, batch_size=32, shuffle=True)  
val_dataloader = DataLoader(val_dataset, batch_size=32, shuffle=False)
```

```python
def train_one_epoch_multiclass(model, dataloader, criterion, optimizer, device):
    model.train()
    sum_loss = 0.0
    sum_acc = 0.0

    for i, (X_batch, y_batch) in enumerate(dataloader, 1):
        X_batch = X_batch.to(device)
        y_batch = y_batch.to(device)
        batch_loss = 0.0
        
        optimizer.zero_grad()
        y_pred = model(X_batch)
        loss = criterion(y_pred, y_batch)
        loss.backward()
        optimizer.step()

        batch_loss += loss.item()
        sum_loss += batch_loss

        # 다중 클래스 정확도 계산
        y_pred_index = y_pred.argmax(dim=1)
        acc = (y_batch == y_pred_index).float().mean().item() * 100
        sum_acc += acc

        if i % 20 == 0:
            print(f"[Batch {i}/{len(dataloader)}] | [Batch Loss: {batch_loss:.4f}] | [Batch Accuracy: {acc:.2f}]")

    avg_loss = sum_loss / len(dataloader)
    avg_acc = sum_acc / len(dataloader)

    return avg_loss, avg_acc
```

```python
@torch.no_grad()
def evaluate_multiclass(model, dataloader, criterion, device):
    model.eval()
    sum_loss = 0.0
    sum_acc = 0.0

    for X_batch, y_batch in dataloader:
        X_batch = X_batch.to(device)
        y_batch = y_batch.to(device)

        y_pred = model(X_batch)
        loss = criterion(y_pred, y_batch)
        sum_loss += loss.item()

        y_pred_index = y_pred.argmax(dim=1)
        acc = (y_batch == y_pred_index).float().mean().item() * 100
        sum_acc += acc

    avg_loss = sum_loss / len(dataloader)
    avg_acc = sum_acc / len(dataloader)

    return avg_loss, avg_acc
```

```python
def train_model_multiclass(model, model_name, train_loader, val_loader, criterion, optimizer, epochs, device, save_base_dir):
    save_dir = os.path.join(save_base_dir, model_name)
    os.makedirs(save_dir, exist_ok=True)

    history = {
        "train_loss": [], 
        "val_loss": [],
        "train_acc": [],
        "val_acc": []
    }
    best_val_acc = 0.0

    # 학습
    print(f"===================== {model_name} 학습 시작 =====================")
    for epoch in range(epochs):
        lr = optimizer.param_groups[0]["lr"]
        train_loss, train_acc = train_one_epoch_multiclass(model, train_loader, criterion, optimizer, device)
        val_loss, val_acc = evaluate_multiclass(model, val_loader, criterion, device)

        print(f"[Epoch: {epoch + 1}/{epochs}] | Train Loss: {train_loss:.4f} | Train Accuracy: {train_acc:.2f}")
        print(f"[Epoch: {epoch + 1}/{epochs}] | Vaildation Loss: {val_loss:.4f} | Vaildation Accuracy: {val_acc:.2f}")

        history["train_loss"].append(train_loss)
        history["val_loss"].append(val_loss)
        history["train_acc"].append(train_acc)
        history["val_acc"].append(val_acc)

        if val_acc > best_val_acc:
            best_val_acc = val_acc
            torch.save(model.state_dict(), os.path.join(save_dir, "best_model.pth"))

    torch.save(model.state_dict(), os.path.join(save_dir, f"last_model_{epochs}.pth"))
    print(f"===================== {model_name} 학습 종료 =====================")
    return history
```

## Grad-CAM

```python
# ===== [보너스] Grad-CAM : 대상 선별 (정분류 3장 + 오분류 2장) =====
res_model.eval()
val_records = []
offset = 0

with torch.no_grad():
    for batch_imgs, batch_labels in val_dataloader:
        probs_b = F.softmax(res_model(batch_imgs.to(device)), dim=1).cpu()
        preds_b = probs_b.argmax(dim=1)
        for j in range(len(batch_labels)):
            # val_dataloader가 shuffle=False이므로 순번이 val_idx와 1:1 대응
            val_records.append({
                "path": base_dataset.samples[val_idx[offset + j]][0],
                "true": int(batch_labels[j]),
                "pred": int(preds_b[j]),
                "prob": float(probs_b[j][preds_b[j]]),
            })
        offset += len(batch_labels)

# 확신도가 높은 순으로 뽑아야 히트맵의 근거가 뚜렷하게 드러난다
correct_recs = sorted([r for r in val_records if r["true"] == r["pred"]],
                      key=lambda r: r["prob"], reverse=True)
wrong_recs   = sorted([r for r in val_records if r["true"] != r["pred"]],
                      key=lambda r: r["prob"], reverse=True)

targets = correct_recs[:3] + wrong_recs[:2]
print(f"정분류 {len(correct_recs)}건 / 오분류 {len(wrong_recs)}건 → 대상 {len(targets)}장")
```

```python
# ===== 훅 등록 =====
activations = None
gradients = None

def forward_hook(module, input, output):
    global activations
    activations = output.detach()

def backward_hook(module, grad_input, grad_output):
    global gradients
    gradients = grad_output[0].detach()

# backbone이 동결되어 있으면 역전파가 layer4까지 도달하지 않아 gradients가 None으로 남는다.
# 가중치를 갱신하는 게 아니라 기울기 '경로'만 여는 것이므로 파인튜닝이 아니다.
for p in res_model.parameters():
    p.requires_grad_(True)

# ResNet은 블록이 중첩돼 있어 .modules()로 재귀 순회해야 layer4 내부까지 도달한다
conv_layers = [m for m in res_model.modules() if isinstance(m, nn.Conv2d)]
target_layer = conv_layers[-1]          # layer4[1].conv2, Conv2d(512, 512) → 출력 7x7
print(f"타겟 레이어: {target_layer}")

h_fwd = target_layer.register_forward_hook(forward_hook)
h_bwd = target_layer.register_full_backward_hook(backward_hook)
```

```python
# ===== Grad-CAM 계산 + 정량 지표 =====
ROW_NAME = ["상단", "상단", "중앙", "중앙", "중앙", "하단", "하단"]
COL_NAME = ["좌측", "좌측", "중앙", "중앙", "중앙", "우측", "우측"]

def grad_cam(img_path):
    original_image = Image.open(img_path).convert("RGB")
    input_tensor = val_transform(original_image).unsqueeze(0).to(device)

    res_model.eval()
    res_model.zero_grad()

    # Grad-CAM은 역전파가 필요하므로 no_grad를 쓰지 않는다
    output = res_model(input_tensor)
    probs = F.softmax(output, dim=1)[0]
    pred_idx = int(torch.argmax(probs))
    confidence = float(probs[pred_idx]) * 100

    # 2순위 클래스까지 확인하면 '얼마나 아슬아슬하게 갈렸는지'를 알 수 있다
    top2 = torch.topk(probs, 2)
    second_idx = int(top2.indices[1])
    second_conf = float(top2.values[1]) * 100

    output[0, pred_idx].backward()

    pooled_gradients = torch.mean(gradients, dim=[0, 2, 3])                  # (512,)
    weighted_activations = activations[0] * pooled_gradients[:, None, None]  # (512, 7, 7)

    heatmap = torch.mean(weighted_activations, dim=0).cpu().numpy()          # (7, 7)
    heatmap = np.maximum(heatmap, 0)          # ReLU : 양수 기여도만
    if heatmap.max() != 0:
        heatmap /= heatmap.max()

    # --- 관찰의 객관적 근거가 될 정량 지표 ---
    total = heatmap.sum()
    pr, pc = np.unravel_index(heatmap.argmax(), heatmap.shape)   # 최고 집중 지점
    center_ratio = heatmap[2:5, 2:5].sum() / total if total > 0 else 0   # 중앙 3x3 비중

    border = np.ones_like(heatmap, dtype=bool)
    border[1:6, 1:6] = False                                      # 바깥 테두리 한 겹
    border_ratio = heatmap[border].sum() / total if total > 0 else 0

    stats = {
        "peak": f"{ROW_NAME[pr]} {COL_NAME[pc]} (격자 {pr},{pc})",
        "center_ratio": center_ratio,
        "border_ratio": border_ratio,
        "second": (second_idx, second_conf),
    }

    heatmap_pil = Image.fromarray(np.uint8(255 * heatmap)).resize(
        original_image.size, Image.Resampling.BILINEAR
    )
    return original_image, heatmap_pil, pred_idx, confidence, stats
```

```python
# ===== 시각화 (왼쪽 원본 / 오른쪽 overlay) =====
report_dir = os.path.join(base_path, "reports")
os.makedirs(report_dir, exist_ok=True)

fig, axes = plt.subplots(len(targets), 2, figsize=(10, 5 * len(targets)))

for row, rec in enumerate(targets):
    orig, hm, pred_idx, conf, st = grad_cam(rec["path"])

    true_name = class_dict[rec["true"]]
    pred_name = class_dict[pred_idx]
    is_correct = (rec["true"] == pred_idx)
    mark = "정분류" if is_correct else "오분류"

    axes[row, 0].imshow(orig)
    axes[row, 0].set_title(f"[{mark}] 원본\nTrue: {true_name}", fontsize=11, color="blue")
    axes[row, 0].axis("off")

    # 원본 위에 히트맵을 alpha=0.4로 겹친다. jet: 빨강=집중, 파랑=무시
    axes[row, 1].imshow(orig)
    axes[row, 1].imshow(hm, cmap="jet", alpha=0.4)
    axes[row, 1].set_title(f"Grad-CAM Overlay\nPred: {pred_name} ({conf:.1f}%)",
                           fontsize=11, color="green" if is_correct else "red")
    axes[row, 1].axis("off")

    # 리포트에 쓸 관찰 근거 출력
    print(f"\n[{row+1}] {mark} | True: {true_name} → Pred: {pred_name} ({conf:.1f}%)")
    print(f"    파일        : {os.path.basename(rec['path'])}")
    print(f"    최고 집중   : {st['peak']}")
    print(f"    중앙 집중도 : {st['center_ratio']*100:.1f}%  (물체 집중 여부)")
    print(f"    테두리 비중 : {st['border_ratio']*100:.1f}%  (배경 편향 여부)")
    print(f"    2순위 클래스: {class_dict[st['second'][0]]} ({st['second'][1]:.1f}%)")

plt.tight_layout()
plt.savefig(os.path.join(report_dir, "gradcam.png"), dpi=150, bbox_inches="tight")
plt.show()

h_fwd.remove()
h_bwd.remove()
```
